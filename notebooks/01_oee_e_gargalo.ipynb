# Eficiencia Operacional e OEE Industrial

Producao diaria simulada de 4 maquinas, 120 dias cada. Detalhes em `data/README.md`.

```python
import sys
sys.path.append('..')
import matplotlib.pyplot as plt
from src.data.load_data import pipeline_completo
from src.models.oee import calcular_oee, identificar_gargalo

df = pipeline_completo()
df_oee = calcular_oee(df, tempo_planejado_min=480)
df_oee.head()
```

```python
resumo = identificar_gargalo(df_oee)
resumo
```

```python
fig, ax = plt.subplots()
resumo[['disponibilidade', 'performance', 'qualidade']].plot(kind='bar', ax=ax)
ax.set_title('Componentes de OEE por maquina')
ax.set_ylim(0, 1)
plt.tight_layout()
plt.savefig('../reports/figures/oee_componentes.png', dpi=150)
plt.show()
```

## Conclusoes

Cada maquina tem um gargalo diferente: M3 perde principalmente em performance (velocidade abaixo do ideal), M4 em qualidade (mais peca defeituosa), M2 em disponibilidade (mais parada), M1 tambem em performance mas com OEE geral mais alto. Isso da uma prioridade clara de onde investir manutencao ou ajuste de processo primeiro (M3, menor OEE geral).
